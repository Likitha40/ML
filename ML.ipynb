```python
#-----------------Information Gain------------------------
```

```python
import pandas as pd
import numpy as np
from collections import Counter
```

```python
df = pd.read_csv("electronic_customer.csv")
df = df.drop("RID", axis=1)
for col in df.columns:
    df[col] = pd.factorize(df[col])[0]
```

```python
def entropy(y):
    c = Counter(y)
    e = 0
    for v in c.values():
        p = v / len(y)
        e -= p * np.log2(p)
    return e
```

```python
def info_gain(X, y, f):
    parent = entropy(y)
    child = 0
    for val in np.unique(X[:, f]):
        idx = np.where(X[:, f] == val)[0]
        child += (len(idx) / len(y)) * entropy(y[idx])
    return parent - child
```

```python
def build_tree(X, y, features):
    if len(set(y)) == 1:
        return y[0]
    if len(features) == 0:
        return Counter(y).most_common(1)[0][0]
    gains = [info_gain(X, y, f) for f in features]
    best = features[np.argmax(gains)]
    tree = {df.columns[best]: {}}
    for val in np.unique(X[:, best]):
        idx = np.where(X[:, best] == val)[0]
        rem = [f for f in features if f != best]
        tree[df.columns[best]][val] = build_tree(X[idx], y[idx], rem)
    return tree
```

```python
def print_tree(tree, indent=""):
    if not isinstance(tree, dict):
        print(indent + "Class ->", "Yes" if tree == 1 else "No")
        return

    for feature, branches in tree.items():
        print(indent + feature)
        for value, subtree in branches.items():
            print(indent + "├──", value)
            print_tree(subtree, indent + "    ")
```

```python
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
```

```python
tree = build_tree(X, y, list(range(X.shape[1])))
```

```python
print("Decision Tree:")
print_tree(tree)
```

```python
# -----------------------Gini Index--------------------------
```

```python
import pandas as pd
import numpy as np
from collections import Counter
```

```python
df = pd.read_csv("electronic_customer.csv")
df = df.drop("RID", axis=1)
```

```python
# Encode categorical values
for col in df.columns:
    df[col] = pd.factorize(df[col])[0]
```

```python
# Gini Index
def gini(y):
    c = Counter(y)
    g = 1
    for v in c.values():
        p = v / len(y)
        g -= p**2
    return g
```

```python
# Best feature using Gini
def gini_split(X, y, f):
    g = 0
    for val in np.unique(X[:, f]):
        idx = np.where(X[:, f] == val)[0]
        g += (len(idx) / len(y)) * gini(y[idx])
    return g
```

```python
# Build Tree
def build_tree(X, y, features):
    if len(set(y)) == 1:
        return y[0]

    if len(features) == 0:
        return Counter(y).most_common(1)[0][0]

    scores = [gini_split(X, y, f) for f in features]
    best = features[np.argmin(scores)]

    tree = {df.columns[best]: {}}

    for val in np.unique(X[:, best]):
        idx = np.where(X[:, best] == val)[0]
        rem = [f for f in features if f != best]
        tree[df.columns[best]][val] = build_tree(X[idx], y[idx], rem)

    return tree
```

```python
# Print Tree
def print_tree(tree, indent=""):
    if not isinstance(tree, dict):
        print(indent + "Class ->", "Yes" if tree == 1 else "No")
        return

    for feature, branches in tree.items():
        print(indent + feature)
        for value, subtree in branches.items():
            print(indent + "├──", value)
            print_tree(subtree, indent + "    ")
```

```python
# Features and target
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
```

```python
tree = build_tree(X, y, list(range(X.shape[1])))
```

```python
print("Decision Tree:")
print_tree(tree)
```

```python
import numpy as np

class GaussianNaiveBayes:

    def fit(self, X, y):
        self.classes = np.unique(y)
        self.mean = {}
        self.var = {}
        self.prior = {}

        for c in self.classes:
            X_c = X[y == c]
            self.mean[c] = np.mean(X_c, axis=0)
            self.var[c] = np.var(X_c, axis=0)
            self.prior[c] = len(X_c) / len(X)

    def gaussian_probability(self, class_idx, x):
        mean = self.mean[class_idx]
        var = self.var[class_idx]
        var = np.where(var == 0, 1e-9, var)

        numerator = np.exp(-((x - mean) ** 2) / (2 * var))
        denominator = np.sqrt(2 * np.pi * var)

        return numerator / denominator

    def predict(self, X):
        predictions = []

        for x in X:
            posteriors = []

            for c in self.classes:
                prior = np.log(self.prior[c])
                likelihood = np.sum(np.log(self.gaussian_probability(c, x)))
                posteriors.append(prior + likelihood)

            predictions.append(self.classes[np.argmax(posteriors)])

        return np.array(predictions)
```
